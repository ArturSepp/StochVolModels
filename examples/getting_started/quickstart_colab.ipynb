# Log-normal stochastic-volatility quickstart

This notebook runs the authoritative `stochvolmodels` example for the Karasinski-Sepp log-normal stochastic-volatility model with quadratic drift. It prices a European call and a small option chain with `LogSvParams` and `LogSVPricer`, then reports model prices and implied volatilities.

The calculation itself is offline. Network access is used only to install the released wheel and retrieve the matching tagged quickstart source.

## Load the version-matched LogSV example

The notebook reads the installed version and fetches `quickstart.py` from the corresponding GitHub release tag. This keeps the script, documentation, CI, and Colab workflow on one implementation.

```python
from importlib.metadata import version
from pathlib import Path
from urllib.request import urlopen

from IPython.display import Code, display

release = version("stochvolmodels")
quickstart_url = (
    "https://raw.githubusercontent.com/ArturSepp/StochVolModels/"
    f"v{release}/examples/getting_started/quickstart.py"
)
with urlopen(quickstart_url, timeout=30) as response:
    source = response.read().decode("utf-8")

assert "LogSvParams" in source and "LogSVPricer" in source
Path("quickstart.py").write_text(source, encoding="utf-8")
print(f"Loaded the LogSV quickstart for stochvolmodels {release}")
display(Code(source, language="python"))
```

## Price with the LogSV model

```python
%run quickstart.py
```
